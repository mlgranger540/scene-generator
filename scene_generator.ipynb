```python
import csv
from random import *

# column 1 = actions
# column 2 = actions with char
# column 3 = descriptors
# column 4 = reasons
# column 5 = reasons with char
# column 6 = conversation topics
# column 7 = locations
# column 8 = items
# column 9 = uses

# Make lists for each column in CSV
actions=[]
actionswchar=[]
descriptors=[]
reasons=[]
reasonswchar=[]
topics=[]
locations=[]
items=[]
uses=[]

# Open and read CSV file
with open('scene_idea_bits.csv', 'r') as csvfile:
    sceneideabits = csv.reader(csvfile)
    # Skip header row
    next(sceneideabits)
    # Add data from each column of CSV to lists
    for row in sceneideabits:
        action=row[0]
        actionwchar=row[1]
        descriptor=row[2]
        reason=row[3]
        reasonwchar=row[4]
        topic=row[5]
        location=row[6]
        item=row[7]
        use=row[8]
        actions.append(action)
        actionswchar.append(actionwchar)
        descriptors.append(descriptor)
        reasons.append(reason)
        reasonswchar.append(reasonwchar)
        topics.append(topic)
        locations.append(location)
        items.append(item)
        uses.append(use)

# Remove empty rows
while '' in actions:
    actions.remove('')

while '' in actionswchar:
    actionswchar.remove('')

while '' in descriptors:
    descriptors.remove('')

while '' in reasons:
    reasons.remove('')

while '' in reasonswchar:
    reasonswchar.remove('')

while '' in topics:
    topics.remove('')

while '' in locations:
    locations.remove('')

while '' in items:
    items.remove('')

while '' in uses:
    uses.remove('')

# Make character list from inputted character names
characters=[]

char1=input('Name of character 1:')
char2=input('Name of character 2:')
char3=input('Name of character 3:')
char4=input('Name of character 4:')
char5=input('Name of character 5:')
char6=input('Name of character 6:')

characters.append(char1)
characters.append(char2)
characters.append(char3)
characters.append(char4)
characters.append(char5)
characters.append(char6)

# Remove empty character names (if input is left blank)
while '' in characters:
    characters.remove('')

# Make randomised number lists without duplicates (to avoid character names being repeated in first section i.e. Ben was betrayed by Ben xD)
randomunique1 = sample(range(len(characters)), 2)
randomunique2 = sample(range(len(characters)), 2)
randomunique3 = sample(range(len(characters)), 2)
randomunique4 = sample(range(len(characters)), 2)
randomunique5 = sample(range(len(characters)), 2)
randomunique6 = sample(range(len(characters)), 2)
randomunique7 = sample(range(len(characters)), 2)
randomunique8 = sample(range(len(characters)), 2)
randomunique9 = sample(range(len(characters)), 2)
randomunique10 = sample(range(len(characters)), 2)
randomunique11 = sample(range(len(characters)), 2)
randomunique12 = sample(range(len(characters)), 2)
randomunique13 = sample(range(len(characters)), 2)
randomunique14 = sample(range(len(characters)), 2)
randomuniquethree1 = sample(range(len(characters)), 3)
randomuniquethree2 = sample(range(len(characters)), 3)
randomuniqueitems1 = sample(range(len(items)), 2)


# Print randomised prompts using character names, actions, descriptors, reasons, locations, items and uses
print('')
# Single character prompts
# char + action
print(choice(characters),choice(actions),end='.\n')
# char + descriptor
print(choice(characters),'is',choice(descriptors),end='.\n')
# char + descriptor + location
print(choice(characters),'is',choice(descriptors),choice(locations),end='.\n')
# char + location + item
print(choice(characters),'is',choice(locations),'with nothing but',choice(items),end='.\n')

print('')
# Multiple character prompts
# char + action + char
print(characters[randomunique1[0]],choice(actionswchar),characters[randomunique1[1]],end='.\n')
# char + descriptor + char
print(characters[randomunique2[0]],'is',choice(descriptors),'with',characters[randomunique2[1]],end='.\n')
# char + location + char
print(characters[randomunique3[0]],'is',choice(locations),'with',characters[randomunique3[1]],end='.\n')
# char + descriptor + char + location
print(characters[randomunique4[0]],'is',choice(descriptors),'with',characters[randomunique4[1]],choice(locations),end='.\n')

print('')
# Multiple character prompts with reasons
# char + action + char + connective + reason
print(characters[randomunique5[0]],choice(actionswchar),characters[randomunique5[1]],'because',choice(reasons),end='.\n')
print(characters[randomunique6[0]],choice(actionswchar),characters[randomunique6[1]],'after',choice(reasons),end='.\n')
# char + action + char + connective + char + reason
print(characters[randomunique7[0]],choice(actionswchar),characters[randomunique7[1]],'because',choice(characters),choice(reasonswchar),end='.\n')
print(characters[randomunique8[0]],choice(actionswchar),characters[randomunique8[1]],'after',choice(characters),choice(reasonswchar),end='.\n')

print('')
# Conversation/argument prompts
# char + converses + char(s) + topic
print(characters[randomunique9[0]],'has a conversation with',characters[randomunique9[1]],'about',choice(topics),end='.\n')
print(characters[randomuniquethree1[0]],'has a conversation with',characters[randomuniquethree1[1]],'and',characters[randomuniquethree1[2]],'about',choice(topics),end='.\n')
# char + converses + char + topic + location
print(characters[randomunique10[0]],'has a conversation with',characters[randomunique10[1]],'about',choice(topics),choice(locations),end='.\n')
# char + argues + char(s) + topic
print(characters[randomunique11[0]],'has an argument with',characters[randomunique11[1]],'about',topics[randint(1, len(topics)-1)],end='.\n')
print(characters[randomuniquethree2[0]],'has an argument with',characters[randomuniquethree2[1]],'and',characters[randomuniquethree2[2]],'about',topics[randint(1, len(topics)-1)],end='.\n')
# char + argues + char + topic + location
print(characters[randomunique12[0]],'has an argument with',characters[randomunique12[1]],'about',choice(topics),choice(locations),end='.\n')

print('')
# Item prompts
# char + finds + item + somewhere unexpected
print(choice(characters),'finds',choice(items),'somewhere unexpected',end='.\n')
# char + finds + item + location
print(choice(characters),'finds',choice(items),choice(locations),end='.\n')
# char + gives + char + item
print(characters[randomunique13[0]],'gives',characters[randomunique13[1]],choice(items),end='.\n')
# char + steals + item + from + char
print(characters[randomunique14[0]],'steals',choice(items),'from',characters[randomunique14[1]],end='.\n')
# char + uses + item(s) + to + use
print(choice(characters),'uses',items[randint(1, len(items)-1)],'to',choice(uses),end='.\n')
print(choice(characters),'uses',items[randomuniqueitems1[0]],'and',items[randomuniqueitems1[1]],'to',choice(uses),end='.\n')

print('')


# Print to check lists
# print('')
# print('---')
# print('')
# print('There are', len(actions), 'items in Actions:')
# print(', '.join(actions))
# print('')
# print('There are', len(actionswchar), 'items in Actions with Char:')
# print(', '.join(actionswchar))
# print('')
# print('There are', len(descriptors), 'items in Descriptors:')
# print(', '.join(descriptors))
# print('')
# print('There are', len(reasons), 'items in Reasons:')
# print(', '.join(reasons))
# print('')
# print('There are', len(reasonswchar), 'items in Reasons with Char:')
# print(', '.join(reasonswchar))
# print('')
# print('There are', len(topics), 'items in Conversation Topics:')
# print(', '.join(topics))
# print('')
# print('There are', len(locations), 'items in Locations:')
# print(', '.join(locations))
# print('')
# print('There are', len(items), 'items in Items:')
# print(', '.join(items))
# print('')
# print('There are', len(uses), 'items in Uses:')
# print(', '.join(uses))
```
