```python
import csv
from random import *

sceneideabits=csv.reader(open('scene_idea_bits.csv'))

# column 1 = actions
# column 2 = actions with char
# column 3 = reasons
# column 4 = reasons with char
# column 5 = items
# column 6 = uses
# column 7 = conversations

# Make lists for each column and add data from CSV
actions=[]
actionswchar=[]
reasons=[]
reasonswchar=[]
items=[]
uses=[]
conversations=[]

for row in sceneideabits:
    action=row[0]
    actionwchar=row[1]
    reason=row[2]
    reasonwchar=row[3]
    item=row[4]
    use=row[5]
    conversation=row[6]
    actions.append(action)
    actionswchar.append(actionwchar)
    reasons.append(reason)
    reasonswchar.append(reasonwchar)
    items.append(item)
    uses.append(use)
    conversations.append(conversation)

# Remove empty rows
while '' in actions:
    actions.remove('')

while '' in actionswchar:
    actionswchar.remove('')

while '' in reasons:
    reasons.remove('')

while '' in reasonswchar:
    reasonswchar.remove('')

while '' in items:
    items.remove('')

while '' in uses:
    uses.remove('')

while '' in conversations:
    conversations.remove('')

# Make character list from inputted character names
characters=[]

char1=input('Name of character 1:')
char2=input('Name of character 2:')
char3=input('Name of character 3:')
char4=input('Name of character 4:')
char5=input('Name of character 5:')
char6=input('Name of character 6:')

characters.append(char1)
characters.append(char2)
characters.append(char3)
characters.append(char4)
characters.append(char5)
characters.append(char6)

# Remove empty character names (if input is left blank)
while '' in characters:
    characters.remove('')

# Make randomised number lists without duplicates (to avoid character names being repeated in first section i.e. Ben was betrayed by Ben xD)
randomunique1 = sample(range(len(characters)), 2)
randomunique2 = sample(range(len(characters)), 2)
randomunique3 = sample(range(len(characters)), 2)
randomunique4 = sample(range(len(characters)), 2)
randomunique5 = sample(range(len(characters)), 2)
randomunique6 = sample(range(len(characters)), 2)
randomunique7 = sample(range(len(characters)), 2)
randomunique8 = sample(range(len(characters)), 3)
randomuniqueitems = sample(range(len(items)), 2)


# Print randomised prompts using character names, actions, reasons, items and uses
print('')
# char + action
print(choice(characters),actions[randint(1, len(actions)-1)],end='.\n')
print(choice(characters),actions[randint(1, len(actions)-1)],end='.\n')
# char + action + char
print(characters[randomunique1[0]],actionswchar[randint(1, len(actionswchar)-1)],characters[randomunique1[1]],end='.\n')
print(characters[randomunique2[0]],actionswchar[randint(1, len(actionswchar)-1)],characters[randomunique2[1]],end='.\n')
print('')
# char + action + char + connective + reason
print(characters[randomunique3[0]],actionswchar[randint(1, len(actionswchar)-1)],characters[randomunique3[1]],'after',reasons[randint(1, len(reasons)-1)],end='.\n')
print(characters[randomunique4[0]],actionswchar[randint(1, len(actionswchar)-1)],characters[randomunique4[1]],'because',reasons[randint(1, len(reasons)-1)],end='.\n')
# char + action + char + connective + char + reason
print(characters[randomunique5[0]],actionswchar[randint(1, len(actionswchar)-1)],characters[randomunique5[1]],'after',choice(characters),reasonswchar[randint(1, len(reasonswchar)-1)],end='.\n')
print(characters[randomunique6[0]],actionswchar[randint(1, len(actionswchar)-1)],characters[randomunique6[1]],'because',choice(characters),reasonswchar[randint(1, len(reasonswchar)-1)],end='.\n')
print('')
# char + converses with + character(s) + about + conversation topic
print(characters[randomunique7[0]],'has a conversation with',characters[randomunique7[1]],'about',conversations[randint(1, len(conversations)-1)],end='.\n')
print(characters[randomunique8[0]],'has a conversation with',characters[randomunique8[1]],'and',characters[randomunique8[2]],'about',conversations[randint(1, len(conversations)-1)],end='.\n')
print('')
# char + uses + item(s) + to + use
print(choice(characters),'uses',items[randint(1, len(items)-1)],'to',uses[randint(1, len(uses)-1)],end='.\n')
print(choice(characters),'uses',items[randomuniqueitems[0]],'and',items[randomuniqueitems[1]],'to',uses[randint(1, len(uses)-1)],end='.\n')


# Print to check lists
print('')
print('')
print('')
print('There are', len(actions)-1, 'items in Actions:')
print(', '.join(actions[1:len(actions)]))
print('')
print('There are', len(actionswchar)-1, 'items in Actions with Char:')
print(', '.join(actionswchar[1:len(actionswchar)]))
print('')
print('There are', len(reasons)-1, 'items in Reasons:')
print(', '.join(reasons[1:len(reasons)]))
print('')
print('There are', len(reasonswchar)-1, 'items in Reasons with Char:')
print(', '.join(reasonswchar[1:len(reasonswchar)]))
print('')
print('There are', len(items)-1, 'items in Items:')
print(', '.join(items[1:len(items)]))
print('')
print('There are', len(uses)-1, 'items in Uses:')
print(', '.join(uses[1:len(uses)]))
print('')
print('There are', len(uses)-1, 'items in Conversations:')
print(', '.join(conversations[1:len(conversations)]))
```
